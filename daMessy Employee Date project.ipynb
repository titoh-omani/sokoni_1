**Cleaning**

```python
df['EmploymentStatus'] = df['EmploymentStatus'].str.title().str.strip()

df['EmploymentStatus'].unique()
```

```python
df.columns.tolist()
```

```python
df['HireDate'].head(10)
```

```python
df['HireDate'] = pd.to_datetime(df['HireDate'], errors='coerce', format='mixed')
```

```python
df.head(15)
```

```python
df['Department'].unique()
```

```python
df['City'].unique()
```

```python
df['EmploymentStatus'].unique()
```

```python
df['Age'].unique()
```

```python
#Change age to integer
df["Age"] = df["Age"].replace("thirty", "30")
df['Age'].unique()
```

```python
#Type Conversion for the age to integer 
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df.info()
```

```python
df.info()
```

```python
#Replace Age with Median
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
df.info()
```

```python
df['City'].mode()
```

```python
df['City'].value_counts()
```

```python
df["City"] = df["City"].fillna('unknown')
```

```python
df['City'].unique()
```

```python
df['City'].value_counts()
```

```python
df['Department'] = df['Department'].str.title()
```

```python
df['Department'].unique()
```

```python
df['Department'] = df['Department'].str.replace('It', 'IT').str.replace('Hr', 'HR')
df['Department'].unique()
```

```python
mediansalary = df.groupby('Department')['MonthlySalary'].median()
print(mediansalary)
```

```python
df["MonthlySalary"] = df["MonthlySalary"].fillna(
    df["Department"].map(mediansalary)
)
```

```python
df.info()
```

```python
df["PerformanceScore"].value_counts()
```

```python
# Convert to numeric:
df['PerformanceScore'] = pd.to_numeric(
    df['PerformanceScore'],
    errors='coerce'
)
```

```python
df["PerformanceScore"].value_counts(dropna = False)
```

```python
# fill the missing performance scores with the median:
df["PerformanceScore"] = df["PerformanceScore"].fillna(
    df["PerformanceScore"].median()
)
```

```python
df.info()
```

**Rename Columns**

```python
df.columns = ["employee_id" , "name" , "age" , "department" , "city" , "employment_status" , "monthly_salary" , 
              "perfomance_score" , "hire_date" , "email"]
```

```python
df.head()
```

**Final Quality Check** 

```python
print("Rows after cleaning:", len(df)) 
print("Columns after cleaning:", len(df.columns)) 
print("Completely duplicated rows:", df.duplicated().sum()) 
print("Unique Employee IDs:", df["employee_id"].nunique()) 
print("Duplicate Employee ID rows:", df["employee_id"].duplicated(keep=False).sum()) 
print("Duplicate email rows:", df["email"].duplicated(keep=False).sum()) 
df.info()
```

```python
df["city"] = df["city"].astype("string").str.strip() 
df["EmploymentStatus"] = (df["employment_status"].astype(str).str.strip().str.title()) 
df["name"] = df["name"].astype(str).str.strip().str.title() 
# Simple email cleaning only 
df["email"] = df["email"].astype(str).str.strip().str.lower() 
print("departments:", df["department"].unique()) 
print("employment_statuses:", df["employment_status"].unique()) 
print("cities:", df["city"].dropna().unique())
```

**Save the DataSet**

```python
output_file = "cleaned_employee_data.csv"
df.to_csv(output_file, index=False)
print(f" Cleaned dataset saved as: {output_file}")
```
